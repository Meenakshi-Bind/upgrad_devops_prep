```python
import numpy as np

# -----------------------------
# Load Cork rainfall data
# -----------------------------
cork = np.loadtxt("CorkRainfall.txt")

# Column indices for readability
YEAR = 0
MONTH = 1
TOTAL = 2
MOST_DAY = 3
RAINDAYS = 4

# ---------------------------------------------------------
# (i) Max and Average of “Most Rainfall in a Day”
# ---------------------------------------------------------
max_most_day = np.max(cork[:, MOST_DAY])
avg_most_day = np.mean(cork[:, MOST_DAY])

print("Max 'Most Rainfall in a Day':", max_most_day)
print("Average 'Most Rainfall in a Day':", avg_most_day)

# ---------------------------------------------------------
# (ii) Unique years + Rain days per month for a chosen year
# ---------------------------------------------------------
years = np.unique(cork[:, YEAR]).astype(int)
print("Years available:", years)

selected_year = int(input("Enter a year from the list above: "))

# Filter rows for that year
rows_year = cork[cork[:, YEAR] == selected_year]

# Sum rain days per month
rain_days_per_month = rows_year[:, RAINDAYS]

print(f"Rain days per month in {selected_year}:")
print(rain_days_per_month)

# ---------------------------------------------------------
# (iii) Wettest month (highest cumulative Total Rainfall)
# ---------------------------------------------------------
# Sum rainfall for each month (1–12)
monthly_totals = np.zeros(12)

for m in range(1, 13):
    monthly_totals[m-1] = np.sum(cork[cork[:, MONTH] == m][:, TOTAL])

wettest_month = np.argmax(monthly_totals) + 1
wettest_value = monthly_totals[wettest_month - 1]

print("Wettest month:", wettest_month)
print("Cumulative rainfall:", wettest_value)

# ---------------------------------------------------------
# (iv) Percentage of rows with RainDays <= threshold
# ---------------------------------------------------------
threshold = int(input("Enter maximum rain-days threshold: "))

count = np.sum(cork[:, RAINDAYS] <= threshold)
percentage = (count / len(cork)) * 100

print(f"Percentage of rows with rain days <= {threshold}: {percentage:.2f}%")

# ---------------------------------------------------------
# (v) Average total rainfall for Summer & Autumn
# ---------------------------------------------------------
summer_months = [6, 7, 8]
autumn_months = [9, 10, 11]

summer_avg = np.mean(cork[np.isin(cork[:, MONTH], summer_months)][:, TOTAL])
autumn_avg = np.mean(cork[np.isin(cork[:, MONTH], autumn_months)][:, TOTAL])

print("Average summer rainfall:", summer_avg)
print("Average autumn rainfall:", autumn_avg)

# ---------------------------------------------------------
# (vi) Append Dublin data + average raindays + save to CSV
# ---------------------------------------------------------
dublin = np.loadtxt("DublinRainfall.txt")

combined = np.vstack((cork, dublin))

avg_raindays_combined = np.mean(combined[:, RAINDAYS])
print("Average rain days (Cork + Dublin):", avg_raindays_combined)

np.savetxt("CombinedRainfall.csv", combined, delimiter=",")
print("Combined dataset saved to CombinedRainfall.csv")
```
