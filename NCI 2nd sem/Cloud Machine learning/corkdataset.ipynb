```python
import numpy as np

cork = np.loadtxt("CorkRainfall.txt")

MOST_DAY = 3

max_most_day = np.max(cork[:, MOST_DAY])
avg_most_day = np.mean(cork[:, MOST_DAY])

print("Max Most Rainfall in a Day:", max_most_day)
print("Average Most Rainfall in a Day:", avg_most_day)
```

```python
YEAR = 0
RAINDAYS = 4

years = np.unique(cork[:, YEAR]).astype(int)
print("Available years:", years)

selected_year = int(input("Enter a year: "))

rows = cork[cork[:, YEAR] == selected_year]
rain_days = rows[:, RAINDAYS]

print(f"Rain days per month in {selected_year}:")
print(rain_days)
```

```python
MONTH = 1
TOTAL = 2

monthly_totals = np.zeros(12)

for m in range(1, 13):
    monthly_totals[m-1] = np.sum(cork[cork[:, MONTH] == m][:, TOTAL])

wettest_month = np.argmax(monthly_totals) + 1
wettest_value = monthly_totals[wettest_month - 1]

print("Wettest month:", wettest_month)
print("Cumulative rainfall:", wettest_value)
```

```python
threshold = int(input("Enter rain-days threshold: "))

count = np.sum(cork[:, RAINDAYS] <= threshold)
percentage = (count / len(cork)) * 100

print(f"Percentage of rows with rain days <= {threshold}: {percentage:.2f}%")
```

```python
dublin = np.loadtxt("DublinRainfall.txt")

combined = np.vstack((cork, dublin))

avg_raindays_combined = np.mean(combined[:, RAINDAYS])
print("Average rain days (Cork + Dublin):", avg_raindays_combined)

np.savetxt("CombinedRainfall.csv", combined, delimiter=",")
print("Combined data saved to CombinedRainfall.csv")
```
