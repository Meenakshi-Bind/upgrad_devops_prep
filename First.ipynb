```python
text = input("Enter a string: ")

if text == text[::-1]:
    print("It is a palindrome")
else:
    print("It is not a palindrome")
```

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
def check_number(lst, num):
    if num in lst:
        return True
    else:
        return False

numbers = [1, 3, 5, 7, 9]
print(check_number(numbers, 5))
```

```python
a = [8, 9, 10, 11, 13, 81, 101, 100, 94]

for num in a:
    if num < 2:
        continue

    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if not is_prime:
        continue

    print(num * num)
```
